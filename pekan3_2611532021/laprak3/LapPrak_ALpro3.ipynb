### 2.2.2 Program assignment_2611532021

```python
# nama variabel ditambah 4 digit nim terakhir contoh: angka1_1234
# program ini menggunakan fungsi input()
# nilai yang dimasukkan akan dikonversi menjadi tipe data integer
# program operator assignment dalam python

angka1_2021 = int(input("Input angka-1: "))
angka2_2021 = int(input("Input angka-2: "))

print("\nNilai awal angka1 = ", angka1_2021)
print("Nilai angka2 = ", angka2_2021)

# assignment biasa
hasil_2021 = angka1_2021
print("\nAssignment biasa (=)")
print("Hasil = ", hasil_2021)

# assignment penambahan
hasil_2021 = angka1_2021
hasil_2021 += angka2_2021
print("\nAssignment penambahan (+=)")
print("Hasil = ", hasil_2021)

# assignment pengurangan
hasil_2021 = angka1_2021
hasil_2021 -= angka2_2021
print("\nAssignment pengurangan (-=)")
print("Hasil = ", hasil_2021)

# assignment perkalian
hasil_2021 = angka1_2021
hasil_2021 *= angka2_2021
print("\nAssignment perkalian (*=)")
print("Hasil = ", hasil_2021)

# assignment pembagian, pembagian bulat, dan sisa bagi
if angka2_2021 != 0:
    hasil_2021 = angka1_2021
    hasil_2021 /= angka2_2021
    print("\nAssignment pembagian (/=)")
    print("Hasil = ", hasil_2021)

    # operator tambahan
    hasil_2021 = angka1_2021
    hasil_2021 //= angka2_2021
    print("\nAssignment pembagian bulat (//=)")
    print("Hasil = ", hasil_2021)

    hasil_2021 = angka1_2021
    hasil_2021 %= angka2_2021
    print("\nAssignment sisa bagi (%=)")
    print("Hasil = ", hasil_2021)
else:
    print("\nPembagian tidak dapat dilakukan.")
    print("Angka kedua tidak boleh bernilai 0.")

# operator tambahan: assignment perpangkatan
hasil_2021 = angka1_2021
hasil_2021 **= angka2_2021
print("\nAssignment perpangkatan (**=)")
print("Hasil = ", hasil_2021)

```

**Penjelasan :**

1. `input()` digunakan untuk menerima masukan dari pengguna, sedangkan `int()` digunakan untuk mengubah masukan tersebut menjadi tipe data integer.

2. `angka1_2021` dan `angka2_2021` merupakan variabel yang digunakan untuk menyimpan dua angka yang dimasukkan oleh pengguna.

3. `print()` digunakan untuk menampilkan nilai awal dari `angka1_2021` dan `angka2_2021`.

4. Operator assignment biasa (`=`) digunakan untuk memberikan nilai `angka1_2021` kepada variabel `hasil_2021`.

5. Operator assignment penambahan (`+=`) digunakan untuk menambahkan nilai `angka2_2021` ke `hasil_2021`. Operasi `hasil_2021 += angka2_2021` setara dengan `hasil_2021 = hasil_2021 + angka2_2021`.

6. Operator assignment pengurangan (`-=`) digunakan untuk mengurangi nilai `hasil_2021` dengan `angka2_2021`. Operasi `hasil_2021 -= angka2_2021` setara dengan `hasil_2021 = hasil_2021 - angka2_2021`.

7. Operator assignment perkalian (`*=`) digunakan untuk mengalikan nilai `hasil_2021` dengan `angka2_2021`. Operasi `hasil_2021 *= angka2_2021` setara dengan `hasil_2021 = hasil_2021 * angka2_2021`.

8. Percabangan `if angka2_2021 != 0` digunakan untuk memastikan bahwa `angka2_2021` tidak bernilai `0` sebelum melakukan operasi pembagian.

9. Operator assignment pembagian (`/=`) digunakan untuk membagi nilai `hasil_2021` dengan `angka2_2021`. Operasi `hasil_2021 /= angka2_2021` setara dengan `hasil_2021 = hasil_2021 / angka2_2021`.

### 2.2.3 Program bitwise_2611532021

```python
# Buat file dengan nama bitwise_NIM.py
# Nama variabel ditambah 4 digit nim terakhir contoh: angka1_1234
# Program ini menggunakan fungsi input()

print("\n=====================================")
print("3. OPERATOR BITWISE")
print("=====================================")

angka1_2021 = int(input("Masukkan angka bitwise-1: "))
angka2_2021 = int(input("Masukkan angka bitwise-2: "))

print("\nAngka dalam bentuk desimal dan biner")
print("angka1 =", angka1_2021, "| biner =", bin(angka1_2021))
print("angka2 =", angka2_2021, "| biner =", bin(angka2_2021))

# Bitwise AND
hasil1_2021 = angka1_2021 & angka2_2021
print("\nBitwise AND (&)")
print(angka1_2021, "&", angka2_2021, "=", hasil1_2021)
print("Biner hasil =", bin(hasil1_2021))
print("Biner hasil (8 bit) =", format(hasil1_2021, "08b"))

# Bitwise OR
hasil1_2021 = angka1_2021 | angka2_2021
print("\nBitwise OR (|)")
print(angka1_2021, "|", angka2_2021, "=", hasil1_2021)
print("Biner hasil =", bin(hasil1_2021))
print("Biner hasil (8 bit) =", format(hasil1_2021, "08b"))

# Bitwise XOR
hasil1_2021 = angka1_2021 ^ angka2_2021
print("\nBitwise XOR (^)")
print(angka1_2021, "^", angka2_2021, "=", hasil1_2021)
print("Biner hasil =", bin(hasil1_2021))
print("Biner hasil (8 bit) =", format(hasil1_2021, "08b"))

# Bitwise NOT
hasil1_2021 = ~angka1_2021
print("\nBitwise NOT (-)")
print("~", angka1_2021, "=", hasil1_2021)
print("Biner hasil =", bin(hasil1_2021))
print("Biner hasil (8 bit) =", format(hasil1_2021, "08b"))

# Bitwise geser kiri
jumlah_geser_2021 = int(input("\nMasukkan jumlah pergeseran bit: "))

hasil1_2021 = angka1_2021 << jumlah_geser_2021
print("\nBitwise geser kiri (<<)")
print(angka1_2021, "<<", jumlah_geser_2021, "=", hasil1_2021)
print("Biner hasil =", bin(hasil1_2021))
print("Biner hasil (8 bit) =", format(hasil1_2021, "08b"))

# Bitwise geser kanan
hasil1_2021 = angka1_2021 >> jumlah_geser_2021
print("\nBitwise geser kanan (>>)")
print(angka1_2021, ">>", jumlah_geser_2021, "=", hasil1_2021)
print("Biner hasil =", bin(hasil1_2021))
print("Biner hasil (8 bit) =", format(hasil1_2021, "08b"))
```

**Penjelasan :**

1. `input()` digunakan untuk menerima masukan dari pengguna, sedangkan `int()` digunakan untuk mengubah masukan tersebut menjadi tipe data integer.

2. `angka1_2021` dan `angka2_2021` digunakan untuk menyimpan dua bilangan yang akan digunakan dalam operasi bitwise.

3. Fungsi `bin()` digunakan untuk mengubah bilangan integer menjadi representasi bilangan biner. Hasil dari `bin()` memiliki awalan `0b` yang menunjukkan bahwa nilai tersebut merupakan bilangan biner.

4. Operator bitwise AND (`&`) membandingkan setiap bit dari dua bilangan. Hasilnya bernilai `1` jika kedua bit yang dibandingkan sama-sama bernilai `1`; jika tidak, hasilnya `0`.

5. Operator bitwise OR (`|`) membandingkan setiap bit dari dua bilangan. Hasilnya bernilai `1` jika salah satu atau kedua bit yang dibandingkan bernilai `1`.

6. Operator bitwise XOR (`^`) membandingkan setiap bit dari dua bilangan. Hasilnya bernilai `1` jika kedua bit yang dibandingkan berbeda dan `0` jika keduanya sama.

7. Operator bitwise NOT (`~`) membalik nilai setiap bit dari suatu bilangan, yaitu bit `1` menjadi `0` dan bit `0` menjadi `1`. Dalam Python, hasil operator `~` pada bilangan integer mengikuti representasi bilangan bertanda (*signed integer*), sehingga hasilnya dapat berupa bilangan negatif.

8. Fungsi `format(hasil1_2021, "08b")` digunakan untuk menampilkan hasil dalam bentuk biner dengan panjang minimal 8 bit. Angka `0` digunakan untuk menambahkan angka nol di bagian depan jika jumlah bit kurang dari 8.

9. `jumlah_geser_2021` digunakan untuk menyimpan jumlah posisi bit yang akan digeser. Nilainya diperoleh dari masukan pengguna dan dikonversi menjadi integer menggunakan `int()`.

10. Operator bitwise geser kiri (`<<`) digunakan untuk menggeser bit suatu bilangan ke arah kiri sebanyak jumlah posisi yang ditentukan oleh `jumlah_geser_2021`. Setiap pergeseran satu posisi ke kiri pada bilangan positif secara umum setara dengan mengalikan bilangan tersebut dengan `2`.

11. Operator bitwise geser kanan (`>>`) digunakan untuk menggeser bit suatu bilangan ke arah kanan sebanyak jumlah posisi yang ditentukan. Pada bilangan positif, setiap pergeseran satu posisi ke kanan secara umum setara dengan pembagian bulat dengan `2`.

12. Variabel `hasil1_2021` digunakan untuk menyimpan hasil dari setiap operasi bitwise sebelum ditampilkan menggunakan `print()`.

13. Fungsi `print()` digunakan untuk menampilkan bilangan dalam bentuk desimal, biner, serta hasil dari setiap operasi bitwise.

### 2.2.5 Program perbandingan_2611532021

```python
# nama variabel ditambah 4 digit nim terakhir contoh: angka1_1234
# program ini menggunakan fungsi input()
# nilai yang dimasukkan akan dikonversi menjadi tipe data integer
# program operator perbandingan dalam python

angka1_2021 = int(input("Input angka-1: : "))
angka2_2021 = int(input("Input angka-2: : "))

# lebih besar dari
hasil_2611532021 = angka1_2021 > angka2_2021
print("\nOperator lebih besar dari")
print("angka1 > angka2 = ", hasil_2611532021)

# lebih kecil dari
hasil_2611532021 = angka1_2021 < angka2_2021
print("\nOperator lebih kecil dari")
print("angka1 < angka2 = ", hasil_2611532021)

# lebih besar dari atau sama dengan
hasil_2611532021 = angka1_2021 >= angka2_2021
print("\nOperator lebih besar dari atau sama dengan")
print("angka1 >= angka2 = ", hasil_2611532021)

# lebih kecil dari atau sama dengan
hasil_2611532021 = angka1_2021 <= angka2_2021
print("\nOperator lebih kecil dari atau sama dengan")
print("angka1 <= angka2 = ", hasil_2611532021)

# sama dengan
hasil_2611532021 = angka1_2021 == angka2_2021
print("\nOperator sama dengan")
print("angka1 == angka2 = ", hasil_2611532021)

# tidak sama dengan
hasil_2611532021 = angka1_2021 != angka2_2021
print("\nOperator tidak sama dengan")
print("angka1 != angka2 = ", hasil_2611532021)

# tambahan: perbandingan berantai dalam python
hasil_2611532021 = 0 < angka1_2021 < 100
print("\nPerbandingan berantai")
print("0 < angka1 < 100 = ", hasil_2611532021)

hasil_2611532021 = 0 < angka2_2021 < 100
print("0 < angka2 < 100 = ", hasil_2611532021)
```

**Penjelasan :**

1. `input()` digunakan untuk menerima masukan dari pengguna, sedangkan `int()` digunakan untuk mengonversi masukan tersebut menjadi tipe data integer.

```python
# Tugas 3
# Sistem Simulasi Transaksi dan Validasi Akses Toko


print("=== SISTEM TRANSAKSI TOKO ===")

# ==============================
# INPUT DATA
# ==============================

nama_2021 = input("Masukkan Nama Pelanggan : ")

status_2021 = input(
    "Masukkan Status Pelanggan (member/nonmember) : "
).lower()

total_belanja_2021 = int(
    input("Masukkan Total Belanja : ")
)

jumlah_barang_2021 = int(
    input("Masukkan Jumlah Barang : ")
)

kode_promo_2021 = input(
    "Masukkan Kode Promo : "
).upper()


# ==============================
# DATA TRANSAKSI
# ==============================

print("\n=== DATA TRANSAKSI ===")

print("Nama Pelanggan   :", nama_2021)
print("Status Pelanggan :", status_2021)
print("Total Belanja    : Rp", total_belanja_2021)
print("Jumlah Barang    :", jumlah_barang_2021)
print("Kode Promo       :", kode_promo_2021)


# ==============================
# OPERATOR PERBANDINGAN
# ==============================

min_belanja_2021 = total_belanja_2021 >= 200000

min_barang_2021 = jumlah_barang_2021 >= 3

status_member_2021 = status_2021 == "member"


# ==============================
# OPERATOR KEANGGOTAAN
# ==============================

daftar_promo_2021 = [
    "HEMAT10",
    "HEMAT20",
    "GRATISONGKIR"
]

promo_tersedia_2021 = kode_promo_2021 in daftar_promo_2021


# ==============================
# OPERATOR LOGIKA
# ==============================

diskon_member_2021 = (
    status_member_2021
    and min_belanja_2021
)

mendapatkan_promo_2021 = (
    min_barang_2021
    and promo_tersedia_2021
)


print("\n=== HASIL VALIDASI ===")

print(
    "Belanja >= Rp200000       :",
    min_belanja_2021
)

print(
    "Jumlah Barang >= 3        :",
    min_barang_2021
)

print(
    "Status Member             :",
    status_member_2021
)

print(
    "Kode Promo Tersedia       :",
    promo_tersedia_2021
)

print(
    "Mendapatkan Diskon        :",
    diskon_member_2021
)

print(
    "Mendapatkan Promo         :",
    mendapatkan_promo_2021
)


# ==============================
# OPERATOR ARITMATIKA
# ==============================

if diskon_member_2021:
    diskon_2021 = total_belanja_2021 * 10 / 100
else:
    diskon_2021 = 0

total_pembayaran_2021 = (
    total_belanja_2021 - diskon_2021
)

harga_rata_rata_2021 = (
    total_belanja_2021 / jumlah_barang_2021
)

sisa_pembagian_2021 = (
    total_belanja_2021 % jumlah_barang_2021
)


print("\n=== HASIL PERHITUNGAN ===")

print(
    "Diskon                 : Rp",
    diskon_2021
)

print(
    "Total Pembayaran       : Rp",
    total_pembayaran_2021
)

print(
    "Rata-rata Harga Barang : Rp",
    harga_rata_rata_2021
)

print(
    "Sisa Pembagian         :",
    sisa_pembagian_2021
)


# ==============================
# HAK AKSES
# ==============================

member_access_2021 = (
    status_member_2021
    and min_belanja_2021
)

promo_access_2021 = (
    promo_tersedia_2021
    or mendapatkan_promo_2021
)

free_shipping_access_2021 = (
    kode_promo_2021 == "GRATISONGKIR"
    and not status_member_2021
)


print("\n=== HAK AKSES PELANGGAN ===")

print(
    "Member Access        :",
    member_access_2021
)

print(
    "Promo Access         :",
    promo_access_2021
)

print(
    "Free Shipping Access :",
    free_shipping_access_2021
)


# ==============================
# OPERATOR PENUGASAN
# ==============================

poin_2021 = 0

poin_2021 += 10

print("\n=== OPERATOR PENUGASAN ===")

print(
    "Poin setelah += 10 :",
    poin_2021
)

poin_2021 -= 2

print(
    "Poin setelah -= 2  :",
    poin_2021
)

poin_2021 *= 2

print(
    "Poin setelah *= 2  :",
    poin_2021
)

poin_2021 /= 2

print(
    "Poin setelah /= 2  :",
    poin_2021
)


# ==============================
# OPERATOR IDENTITAS
# ==============================

objek_a_2021 = ["HEMAT10"]

objek_b_2021 = ["HEMAT10"]

print("\n=== OPERATOR IDENTITAS ===")

print(
    "objek_a == objek_b     :",
    objek_a_2021 == objek_b_2021
)

print(
    "objek_a is objek_b     :",
    objek_a_2021 is objek_b_2021
)

print(
    "objek_a is not objek_b :",
    objek_a_2021 is not objek_b_2021
)


# ==============================
# OPERATOR BITWISE
# ==============================

print("\n=== OPERASI BITWISE ===")

# Nilai bit
bit_member_2021 = 0b0001
bit_belanja_2021 = 0b0010
bit_barang_2021 = 0b0100
bit_promo_2021 = 0b1000


# OR (|)
kode_status_2021 = 0

if status_member_2021:
    kode_status_2021 |= bit_member_2021

if min_belanja_2021:
    kode_status_2021 |= bit_belanja_2021

if min_barang_2021:
    kode_status_2021 |= bit_barang_2021

if promo_tersedia_2021:
    kode_status_2021 |= bit_promo_2021


print("\n=== KODE STATUS TRANSAKSI ===")

print("0001 | 0010 | 0100 | 1000")

print(
    "Kode Biner   :",
    format(kode_status_2021, "04b")
)

print(
    "Kode Desimal :",
    kode_status_2021
)


# AND (&)
cek_member_2021 = (
    kode_status_2021 & bit_member_2021
)

cek_promo_2021 = (
    kode_status_2021 & bit_promo_2021
)


print("\n=== PEMERIKSAAN STATUS ===")

print("\nCek Member")

print(
    format(kode_status_2021, "04b"),
    "&",
    format(bit_member_2021, "04b")
)

print(
    "Hasil Biner   :",
    format(cek_member_2021, "04b")
)

print(
    "Hasil Desimal :",
    cek_member_2021
)


print("\nCek Promo")

print(
    format(kode_status_2021, "04b"),
    "&",
    format(bit_promo_2021, "04b")
)

print(
    "Hasil Biner   :",
    format(cek_promo_2021, "04b")
)

print(
    "Hasil Desimal :",
    cek_promo_2021
)


# XOR (^)

kode_referensi_2021 = 0b1011

hasil_xor_2021 = (
    kode_status_2021 ^ kode_referensi_2021
)


print("\n=== PERBANDINGAN STATUS ===")

print(
    "Kode Transaksi :",
    format(kode_status_2021, "04b")
)

print(
    "Kode Referensi :",
    format(kode_referensi_2021, "04b")
)

print(
    format(kode_status_2021, "04b"),
    "^",
    format(kode_referensi_2021, "04b")
)

print(
    "Hasil Biner   :",
    format(hasil_xor_2021, "04b")
)

print(
    "Hasil Desimal :",
    hasil_xor_2021
)


# SHIFT

hasil_shift_2021 = kode_status_2021 << 1

print("\n=== SHIFT ===")

print(
    format(kode_status_2021, "04b"),
    "<< 1"
)

print(
    "Hasil Biner   :",
    format(hasil_shift_2021, "b")
)

print(
    "Hasil Desimal :",
    hasil_shift_2021
)


# ==============================
# HASIL OPERATOR
# ==============================

print("\n=== HASIL OPERATOR ===")

print("Operator Aritmatika   : digunakan")
print("Operator Perbandingan : digunakan")
print("Operator Logika       : digunakan")
print("Operator Penugasan    : digunakan")
print("Operator Keanggotaan  : digunakan")
print("Operator Identitas    : digunakan")
print("Operator Bitwise      : digunakan")


print("\n=== SELESAI ===")
```

### 1. Jelaskan operator aritmatika apa saja yang digunakan dan bagaimana operator tersebut membantu menghitung transaksi.

Operator aritmatika yang digunakan dalam program adalah:

* (perkalian) digunakan untuk menghitung besar diskon.
/ (pembagian) digunakan untuk menghitung persentase diskon dan rata-rata harga barang.
- (pengurangan) digunakan untuk menghitung total pembayaran setelah diskon.
% (modulus) digunakan untuk menghitung sisa pembagian total belanja dengan jumlah barang.

Operator tersebut membantu program melakukan perhitungan transaksi secara otomatis, seperti menghitung diskon, total pembayaran, rata-rata harga barang, dan sisa pembagian.

### 2. Jelaskan perbedaan operator and, or, dan not berdasarkan penerapannya pada program.
and digunakan ketika semua kondisi harus terpenuhi. Contohnya, status_member_2021 and min_belanja_2021 digunakan untuk menentukan apakah pelanggan mendapatkan diskon member.
or digunakan ketika setidaknya salah satu kondisi harus terpenuhi. Contohnya, promo_tersedia_2021 or mendapatkan_promo_2021 digunakan untuk menentukan hak akses promo.
not digunakan untuk membalik nilai boolean. Contohnya, not status_member_2021 digunakan untuk memeriksa bahwa pelanggan bukan member.
### 3. Jelaskan fungsi operator in dan not in pada pemeriksaan kode promo.

Operator in digunakan untuk memeriksa apakah suatu nilai terdapat di dalam sebuah koleksi. Pada program, kode_promo_2021 in daftar_promo_2021 digunakan untuk memeriksa apakah kode promo yang dimasukkan pelanggan tersedia dalam daftar promo.

Operator not in digunakan untuk memeriksa apakah suatu nilai tidak terdapat di dalam sebuah koleksi. Operator ini tidak digunakan secara langsung dalam program, tetapi dapat digunakan untuk memeriksa apakah kode promo tidak tersedia.

### 4. Apa perbedaan antara is dan == pada Python?

Operator == digunakan untuk membandingkan nilai atau isi dari dua objek, sedangkan operator is digunakan untuk memeriksa apakah dua variabel merujuk pada objek yang sama.

Pada program:

objek_a_2021 == objek_b_2021

menghasilkan True karena kedua list memiliki isi yang sama.

Sedangkan:

objek_a_2021 is objek_b_2021

menghasilkan False karena kedua list merupakan objek yang berbeda meskipun memiliki isi yang sama.
### 5. Jelaskan bagaimana operator bitwise digunakan untuk menentukan hak akses pelanggan.

Operator bitwise digunakan untuk melakukan operasi pada setiap bit dari bilangan biner. Dalam program ini, operator bitwise dapat digunakan untuk merepresentasikan dan memeriksa hak akses pelanggan berdasarkan nilai biner. Setiap bit dapat mewakili jenis akses tertentu, misalnya akses member, promo, atau gratis ongkir.

Contohnya, operator `&` dapat digunakan untuk memeriksa apakah suatu hak akses tertentu dimiliki oleh pelanggan dengan membandingkan bit yang sesuai. Dengan demikian, operator bitwise dapat membantu menentukan hak akses secara efisien menggunakan kombinasi nilai biner.

### 6. Jelaskan alur program mulai dari input hingga menghasilkan total pembayaran dan status promo.

Alur program dimulai dengan pengguna memasukkan nama, status pelanggan, total belanja, jumlah barang, dan kode promo. Data tersebut kemudian diubah ke format yang sesuai, seperti `lower()` untuk status pelanggan dan `upper()` untuk kode promo.

Selanjutnya, program memeriksa beberapa kondisi, seperti apakah pelanggan merupakan member, apakah total belanja mencapai minimal Rp200.000, apakah jumlah barang minimal 3, dan apakah kode promo tersedia.

Hasil pemeriksaan tersebut digunakan oleh operator logika `and`, `or`, dan `not` untuk menentukan apakah pelanggan mendapatkan diskon, promo, atau akses tertentu. Jika pelanggan memenuhi syarat diskon member, program menghitung diskon sebesar 10% dari total belanja.

Setelah itu, program menghitung total pembayaran dengan mengurangi diskon dari total belanja. Program juga menghitung harga rata-rata per barang dan sisa pembagian total belanja dengan jumlah barang.

Terakhir, program menampilkan hasil transaksi, termasuk total pembayaran, status diskon, status promo, dan hak akses pelanggan.

### 7. Identifikasi minimal satu kondisi dalam program yang menghasilkan True dan satu kondisi yang menghasilkan False.

Contoh kondisi yang menghasilkan `True`:

`status_member_2021 = status_2021 == "member"`

Jika pengguna memasukkan status `"member"`, maka kondisi tersebut menghasilkan `True`.

Contoh kondisi yang menghasilkan `False`:

`min_belanja_2021 = total_belanja_2021 >= 200000`

Jika total belanja kurang dari Rp200.000, maka kondisi tersebut menghasilkan `False`.
