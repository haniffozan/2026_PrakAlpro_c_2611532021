### 2.1 Dasar Teori

Percabangan merupakan salah satu struktur dasar dalam pemrograman yang digunakan untuk menentukan tindakan berdasarkan kondisi tertentu. Dengan percabangan, program dapat menghasilkan keluaran yang berbeda sesuai dengan kondisi yang terpenuhi. Dalam Python, percabangan dapat dilakukan menggunakan if, if-else, if-elif-else, multiple if, dan match-case.

### 2.1.1 Percabangan if

Percabangan if digunakan untuk menjalankan suatu blok kode apabila kondisi yang diberikan bernilai True. Jika kondisi bernilai False, blok kode tersebut tidak akan dijalankan.
```python
if nilai > 80:
    print("Lulus")
```
### 2.1.2 Percabangan if-else

Percabangan if-else digunakan ketika terdapat dua kemungkinan tindakan. Jika kondisi pada if bernilai True, blok if dijalankan. Sebaliknya, jika kondisi bernilai False, blok else dijalankan.
```python
if nilai > 80:
    print("Lulus")
else:
    print("gagal")
```
### 2.1.3 Percabangan if-elif-else

Percabangan if-elif-else digunakan ketika terdapat lebih dari dua kemungkinan kondisi. Python memeriksa kondisi secara berurutan dari if hingga elif. Blok kode dari kondisi pertama yang bernilai True akan dijalankan. Jika tidak ada kondisi yang terpenuhi, blok else akan dijalankan.

```python
if nilai > 80:
    print("Lulus")
elif nilai > 70:
    print("cukup")
else:
    print("gagal")
```

### 2.1.4 Multiple if

Multiple if merupakan penggunaan beberapa pernyataan if yang berdiri secara terpisah. Setiap kondisi diperiksa secara independen sehingga lebih dari satu kondisi dapat bernilai True dan menjalankan blok kodenya masing-masing.
```python
if nilai > 80:
  perintah

if is_student:
  perintah

if kelas == 5:
  perintah
```

Struktur ini dapat digunakan ketika beberapa kondisi dapat terpenuhi secara bersamaan, seperti pemberian beberapa jenis diskon yang dapat ditambahkan.

### 2.1.5 Percabangan match-case

match-case digunakan untuk mencocokkan suatu nilai dengan beberapa pilihan atau pola yang telah ditentukan. Setiap case mewakili kemungkinan nilai tertentu. case _ dapat digunakan sebagai kasus default apabila tidak terdapat case yang sesuai.

```python
match nilai:
case 1:
  print("halo")
case 2:
  print("hola")
case _:
  print("plase choose the available language")
```

### 2.1.6 Operator pada Percabangan

Percabangan menggunakan operator untuk membentuk dan menggabungkan kondisi. Beberapa operator yang digunakan dalam praktikum antara lain:

Operator perbandingan, seperti >, <, >=, ==, dan !=, digunakan untuk membandingkan dua nilai.
Operator logika and, digunakan untuk menggabungkan beberapa kondisi dan menghasilkan True apabila seluruh kondisi bernilai True.
Operator keanggotaan in, digunakan untuk memeriksa apakah suatu nilai terdapat di dalam suatu koleksi, seperti list.

Operator tersebut dapat digunakan bersama struktur percabangan untuk menghasilkan keputusan berdasarkan kondisi yang lebih spesifik.

### 2.2.4 Program multi_if2_2611532021
Program ini digunakan untuk menghitung total diskon belanja berdasarkan jumlah belanja, status member, dan kode promo dengan menggunakan beberapa percabangan `if` yang dapat dijalankan secara bersamaan.


````python
# Buat file dengan nama multi_if2_2611532021.py
# Buat program untuk kondisional If
# Nama variabel ditambah 4 digit nim terakhir contoh ipk_1234
# Program ini menggunakan fungsi input()
# Program Menghitung Diskon belanja

# Input dari user
total_belanja_2021 = float(input("Input Total Belanja (Rp): "))

# Input status member (mengecek apakah user mengetik 'y' atau 'ya')
input_member_2021 = input("Apakah Anda Member? (y/t): ").strip().lower()
is_member_2021 = input_member_2021 in ["y", "ya"]

# Input status kode promo (mengecek apakah user mengetik 'y' atau 'ya')
input_promo_2021 = input("Apakah Kode Promo valid? (y/t): ").strip().lower()
kode_promo_valid_2021 = input_promo_2021 in ["y", "ya"]

total_diskon_persen_2021 = 0

# MULTI-IF terpisah Setiap kondisi diperiksa secara independen
# Diskon bisa ditumpuk jika memenuhi beberapa syarat sekaligus

if total_belanja_2021 > 1000000:
    total_diskon_persen_2021 += 10  # Diskon belanja besar

if is_member_2021:
    total_diskon_persen_2021 += 5  # Diskon member

if kode_promo_valid_2021:
    total_diskon_persen_2021 += 15  # Diskon voucher

# Menghitung nominal diskon dan total bayar
nominal_diskon_2021 = total_belanja_2021 * (total_diskon_persen_2021 / 100) 
total_bayar_2021 = total_belanja_2021 - nominal_diskon_2021

# Output hasil
print("\n--- Rincian Pembayaran ---")
print(f"Total Diskon : {total_diskon_persen_2021}% (Rp {nominal_diskon_2021:,.0f})")
print(f"Total Bayar   : Rp {total_bayar_2021:,.0f}")

print(f"Total Diskon yang anda dapatkan : {total_diskon_persen_2021}%")
# Output: Total diskon yang anda dapatkan : 30% jika belanja > 1 juta, member, dan kode promo valid
````

**Penjelasan:**  

1. `total_belanja_2021 = float(input("Input Total Belanja (Rp): "))` digunakan untuk menerima jumlah belanja dari pengguna dan mengubahnya menjadi tipe data `float`.
2. `input_member_2021 = input(...).strip().lower()` digunakan untuk menerima status member, menghapus spasi dengan `strip()`, dan mengubah input menjadi huruf kecil dengan `lower()`.
3. `is_member_2021 = input_member_2021 in ["y", "ya"]` digunakan untuk memeriksa apakah input pengguna adalah `"y"` atau `"ya"`.
4. `kode_promo_valid_2021 = input_promo_2021 in ["y", "ya"]` digunakan untuk memeriksa apakah pengguna memasukkan jawaban yang menunjukkan bahwa kode promo valid.
5. `total_diskon_persen_2021 = 0` digunakan untuk menentukan nilai awal total diskon sebesar 0%.
6. `if total_belanja_2021 > 1000000:` memberikan tambahan diskon 10% jika total belanja lebih dari Rp1.000.000.
7. `if is_member_2021:` memberikan tambahan diskon 5% jika pengguna merupakan member.
8. `if kode_promo_valid_2021:` memberikan tambahan diskon 15% jika kode promo dinyatakan valid.
9. Operator `+=` digunakan untuk menambahkan setiap diskon ke nilai `total_diskon_persen_2021`, sehingga beberapa diskon dapat ditumpuk.
10. `nominal_diskon_2021 = total_belanja_2021 * (total_diskon_persen_2021 / 100)` digunakan untuk menghitung nominal diskon berdasarkan persentase diskon.
11. `total_bayar_2021 = total_belanja_2021 - nominal_diskon_2021` digunakan untuk menghitung jumlah yang harus dibayar setelah diskon.
12. Fungsi `print()` dan f-string digunakan untuk menampilkan rincian total diskon dan total pembayaran. Jika semua syarat terpenuhi, total diskon yang diperoleh adalah 30%.

## 2.2.7 Program match_case_2611532021.py
Program ini digunakan untuk mengonversi angka 1–12 menjadi nama bulan menggunakan percabangan `match-case`.


````python
# Buat File dengan nama match_case.py
# Buat program untuk match case
# Nama variabel ditambah 4 digit nim terakhir contoh ipk_1234
# Program ini menggunakan fungsi input()
# Program konversi angka menjadi nama bulan

bulan_2021 = int(input("Masukkn Angka bulan (1-12: )"))

match bulan_2021:
    case 1:
        print("Januari")
    case 2:
        print("Februari")
    case 3:
        print("Maret")
    case 4:
        print("April")
    case 5:
        print("Mei")
    case 6:
        print("Juni")
    case 7:
        print("Juli")
    case 8:
        print("Agustus")
    case 9:
        print("September")
    case 10:
        print("Oktober")
    case 11:
        print("November")
    case 12:
        print("Desember")
    case _:
        print("Angka tidak valid")

print("Program selesai")
````

**Penjelasan:**  

1. `bulan_2021 = int(input("Masukkn Angka bulan (1-12: )"))` digunakan untuk menerima angka bulan dari pengguna dan mengubahnya menjadi tipe data `int`.

2. `match bulan_2021:` digunakan untuk mencocokkan nilai `bulan_2021` dengan beberapa pilihan yang tersedia.

3. `case 1:` hingga `case 12:` digunakan untuk mencocokkan angka bulan dengan nama bulan yang sesuai, seperti `1` untuk Januari dan `12` untuk Desember.

4. `case _:` digunakan sebagai kondisi default apabila angka yang dimasukkan tidak sesuai dengan nilai 1–12. Program akan menampilkan `"Angka tidak valid"`.

5. `print("Program selesai")` digunakan untuk menampilkan pesan bahwa program telah selesai dijalankan.

````python
# tugas 4
# menerapkan konsep if, if-else, if-elif-else, multi if, match-case

# 1. masukkan nama pengunjung dan umur
print("=== SISTEM LOKET ALPRO ADVENTURE PARK ===")
nama_pengunjung_2021 = input("Masukkan nama pengunjung: ")
umur_pengunjung_2021 = int(input("Input umur anda: "))
if umur_pengunjung_2021 < 10:
    print("Maaf, syarat umur minimal untuk menikmati wahana adalah 10 tahun.")
    print(f"Umur anda saat ini: {umur_pengunjung_2021} tahun.")
    print("Pembayaran tidak bisa dilanjutkan.")
    exit()

# 2. tanyakan SIM dan cek apakah user memasukkan "ya" atau atau "y"
# tampilkan pilihan paket
sim_2021 = input("Apakah anda sudah punya SIM C (y/t):").strip().lower()[0]
paket_2021 = """
Pilihan Paket Wahana (1-5):
    1. "Wahana Safari Rimba" | Harga Satuan: Rp 50.000
    2. "Wahana Arung Jeram" | Harga Satuan: Rp 75.000
    3. "Wahana Motor ATV Ekstrim" | Harga Satuan: Rp 120.000
    4. "Wahana Roller Coaster Kilat" | Harga Satuan: Rp 100.000
    5. "Wahana All-Access VIP" | Harga Satuan: Rp 220.000"""
print(paket_2021)

# 3. paket yang mana, berapa tiket, member atau tidak, validasi promo
paket_dipilih_2021 = int(input("Masukkan nomor paket (1-5): "))
jumlah_tiket_2021 = int(input("Masukkan jumlah tiket: "))
if jumlah_tiket_2021 <= 0:
    print("Jumlah tiket tidak valid.\nPembayaran tidak bisa dilanjutkan.")
    exit(1) # berhenti jika input user salah

is_member_2021 = input("Apakah Anda Member? (y/t): ").strip().lower()[0]
validasi_promo_2021 = input("Apakah Promo Valid? (y/t): ").strip().lower()[0]

# 4. deklarasi beberapa variabel dan menentukan case untuk tiap wahana
safari_2021 = 50000
arung_2021 = 75000
atv_2021 = 120000
coaster_2021 = 100000
vip_2021 = 220000
subtotal_2021 = 0
umur_minimal_2021 = 17 # umur minimal naik atv

match paket_dipilih_2021:
    case 1: # Safari Rimba 50rb
        subtotal_2021 += safari_2021 * jumlah_tiket_2021

    case 2: # Arung Jeram 75rb
        subtotal_2021 += arung_2021 * jumlah_tiket_2021

    case 3: # motor ATV ekstrim 120rb dan dengan syarat tertentu
        subtotal_2021 += atv_2021 * jumlah_tiket_2021

    case 4: # roller coaster kilat 100rb
        subtotal_2021 += coaster_2021 * jumlah_tiket_2021

    case 5: # All-access VIP
        subtotal_2021 += vip_2021 * jumlah_tiket_2021

    case _:
        print("Paket wahana tidak valid!")
        exit(2)

# 5. menentukan diskon dan harga akhir
total_diskon_2021 = 0

# diskon belanja besar
if subtotal_2021 >= 200:
    total_diskon_2021 += 10

# diskon member
if is_member_2021 == 'y':
    total_diskon_2021 += 5

# diskon kode voucher promo
if validasi_promo_2021 == 'y':
    total_diskon_2021 += 15

# diskon tambahan rombongan
if jumlah_tiket_2021 >= 5:
    total_diskon_2021 += 5

# souvenir
souvenir_2021 = "..."
if subtotal_2021 > 300000:
    souvenir_2021 = "Selamat! Anda berhak mendapatkan Souvenir Gratis."

else:
    souvenir_2021 = "Terima kasih telah berkunjung."

nominal_diskon_2021 = subtotal_2021 * (total_diskon_2021/100)
total_bayar_2021 = subtotal_2021 - nominal_diskon_2021

# 6. syarat untuk atv atau vip (paket vip juga perlu syarat ini karena paket vip memungkinkan
# pembeli menggunakan ATV sehingga pembeli harus memenuhi persyaratan umur dan sim)
if paket_dipilih_2021 == 3 or paket_dipilih_2021 == 5:
    print("\n\n--- KELAYAKAN PENGENDARA WAHANA ---")

if (paket_dipilih_2021 == 3 or paket_dipilih_2021 == 5) and umur_pengunjung_2021 >= 17 and sim_2021 == 'y':
    print("Status  Akses: Anda sudah dewasa dan boleh mengendarai ATV sendiri.")

elif (paket_dipilih_2021 == 3 or paket_dipilih_2021 == 5) and umur_pengunjung_2021 >= 17 and sim_2021 != 'y':
    print("Status  Akses: Anda sudah dewasa tetapi tidak boleh bawa motor ATV (wajib didampingi instruktur)")

elif (paket_dipilih_2021 == 3 or paket_dipilih_2021 == 5) and umur_pengunjung_2021 < 17 and sim_2021 == 'y':
    print("Status  Akses: Identitas tidak valid: Belum cukup umur tapi sudah punya SIM\nPembayaran tidak bisa dilanjutkan.")
    exit(3)

# catatan: berdasarkan instruksi dari codata, seharusnya baris kode dibawah ini adalah `else` dan 
# bukan `elif`. tapi tidak mungkin rasanya kode bisa berjalan dengan benar jika conditional di akhir 
# menggunakan `else` dan bukan `elif`
elif (paket_dipilih_2021 == 3 or paket_dipilih_2021 == 5) and umur_pengunjung_2021 < 17 and sim_2021 == 'y':
    print("Status  Akses: Anda belum cukup umur dan tidak boleh bawa motor ATV.\nPembayaran tidak bisa dilanjutkan.")
    exit(4)

# 7. print rincian pembayaran
print("\n\n--- Rincian Pembayaran ---")
print(f"Subtotal Belanja: Rp {subtotal_2021:,.0f}")
print(f"Total Diskon: {total_diskon_2021}% (Rp {nominal_diskon_2021:,.0f})")
print(f"Total Bayar: Rp {total_bayar_2021:,.0f}")
print("Catatan Layanan: ", souvenir_2021)
print("Program Selesai.")
````
