## 2.1 Dasar Teori

Loop atau perulangan adalah suatu mekanisme digunakan di dunia pemograman salah satunya Python. perulangan memungkinkan suatu baris kode dieksekusi beberapa kali sesuai kebutuhan. ada beberapa macam bentuk loop/perulangan dalam Python tapi pembahasan kali ini akan fokus membahas loop / perulangan jenis "for"

### 2.1.1 For loop/perulangan for

Perulangan for digunakan ketika kita ingin membuat program kita mengeksekusi suatu kode beberapa kali sebanyak yang telah kita tentukan. berikut contoh pengunaannya:

Misal, kita ingin mengeksekusi kode `print("Halo")` dalam Python sebanyak 5 kali maka kita bisa menulisnya seperti ini:
```python
for i in range(5):
    print("Halo")
```
### 2.1.2 Cara menggunakan fungsi `range()` di python

Range biasanya digunakan bersama for loop. Ia biasanya digunakan untuk menentukan batas akhir dari suatu for loop. Secara default, jika kita memasukkan angka 5 dalam fungsi ini maka yang sebenarnya python lakukan adalah menghitung dari 0 sampai 4. Perhatikan bahwa angka 5 tidak termasuk dalam perhitungan walaupun perulangan tetap akan dilakukan sebanyak 5 kali.

Lebih detailnya, seecara bawaan, fungsi range dalam Python menghitung angka dari 0 sampai n - 1 dengan n adalah angka yang dimasukkan user dalam tanda kurung. Lompatan tiap perulangannya juga dilakukan per-1 kali secara bawaan.

Kita bisa mengotak-atik fungsi range ini sehingga ia bekerja seperti yang kita mau. Begini, fungsi `range()` pada Pyhton sebenarnya memiliki format seperti ini `range(mulai dari mana, berhenti sebelum angka ini, berapa banyak penambahannya)` jadi ketika kita menulis `range(1, 6, 1)` perulangan akan dimulai dari 1 taoi perulangan tetap dilakukan sebanyak 5 kali. Jika kita menulis `range(6, 0 , -1)` maka perulangan juga akan terjadi sebanyak lima kali akan tetapi perulangan dimulai dari 6 mundur ke 1 (nol tidak termasuk) yang sama saja dengan perhitungan mundur. contoh:
```python
x = 6
for j in range(x, 0, -1):
    print("urutan ke: ", j)
```

### 2.1.3 nested loop

Nested loop adalah ketika ada suatu loop yang menyimpan loop juga di dalamnya. Dalam hal ini nested for loop bisa berbentuk seperti ini:
```python
x = 7
for i in range(x):
    for j in range(i):
        print("#")
```

### 2.2.7 Program nested_for4_2611532021.py
Program ini menggambar gambar seperti gambar portal di terminal


````python
# Buat file dengan nama nested_for4_NIM.py
# Buat program untuk perulangan for dalam Python
# Nama variabel ditambah 4 digit nim terakhir contoh: ulang_1234
# Program ini menggunakan fungsi input()

tinggi_2021 = int(input("Masukkan tinggi pola (bilangan genap, misal 10): "))

if tinggi_2021 % 2 != 0:
    print("Tinggi harus bilangan genap!")
else:
    a_2021 = tinggi_2021
    c_2021 = a_2021
    lebar_2021 = (2 * tinggi_2021) - 2

    for i_2021 in range(1, tinggi_2021 + 1):
        b_2021 = c_2021 + 1

        for j_2021 in range(1, lebar_2021 + 1):

            # Baris atas dan bawah
            if i_2021 == 1 or i_2021 == tinggi_2021:
                if j_2021 == 1 or j_2021 == lebar_2021:
                    print("#", end="")
                else:
                    print("=", end="")

            # Baris isi
            else:
                if j_2021 == 1 or j_2021 == lebar_2021:
                    print("|", end="")
                else:
                    if j_2021 == c_2021:
                        print("<", end="")
                    elif j_2021 == b_2021:
                        print(">", end="")
                    elif j_2021 == (lebar_2021 - c_2021):
                        print("<", end="")
                    elif j_2021 == (lebar_2021 - c_2021 + 1):
                        print(">", end="")
                    elif j_2021 > b_2021 and j_2021 < (lebar_2021 - c_2021):
                        print(".", end="")
                    else:
                        print(" ", end="")

        print()

        # Logika asli java
        a_2021 -= 2

        if a_2021 <= 0:
            c_2021 = (-a_2021 + 2)
        else:
            c_2021 = a_2021
````

**Penjelasan:**  

1. tinggi_2021 = int(input(...)) digunakan untuk menerima tinggi pola. Program kemudian memeriksa dengan tinggi_2021 % 2 != 0 apakah nilai tersebut bilangan ganjil. Jika ganjil, program menampilkan pesan bahwa tinggi harus bilangan genap.

2. Variabel a_2021, c_2021, dan lebar_2021 digunakan untuk menentukan posisi dan ukuran pola. lebar_2021 = (2 * tinggi_2021) - 2 menentukan jumlah kolom yang digunakan.

3. for i_2021 in range(1, tinggi_2021 + 1): merupakan perulangan luar yang digunakan untuk membuat setiap baris pola, sedangkan for j_2021 in range(1, lebar_2021 + 1): digunakan untuk membuat setiap karakter pada baris tersebut.

4. Pada baris pertama dan terakhir, program menggunakan # pada bagian sudut dan = pada bagian lainnya. Pada baris tengah, program menggunakan |, <, >, ., dan spasi berdasarkan posisi kolom.

5. Percabangan if, elif, dan else digunakan untuk menentukan karakter yang dicetak pada setiap posisi berdasarkan nilai j_2021, b_2021, dan c_2021.

6. Setelah setiap baris selesai dibuat, a_2021 -= 2 dan perhitungan ulang c_2021 digunakan untuk mengubah posisi pola pada baris berikutnya sehingga terbentuk pola yang diinginkan.

7. print() setelah perulangan dalam digunakan untuk berpindah ke baris berikutnya sehingga seluruh karakter membentuk pola dua dimensi.

### 2.2.8 Program jumlah_genap_2611532021.py
Program ini menghitung jumlah penambahan seluruh angka genap dari 2 sampai input

````python
# Buat file dengan nama jumlah_genap_NIM.py
# Buat program untuk perulangan for dalam Python
# Nama variabel ditambah 4 digit nim terakhir contoh: ulang_1234
# Program ini menggunakan fungsi input()

ulang_2021 = int(input("Masukan nilai batas: "))

jumlah_2021 = 0
for i_2021 in range(1, ulang_2021 + 1):
    if i_2021 % 2 == 0:
        print(i_2021, end=" ")
        jumlah_2021 = jumlah_2021 + i_2021

        if i_2021 < ulang_2021:
            print(" + ", end="")
        else:
            print(" = ", jumlah_2021, end="")
print()
print("Jumlah =", jumlah_2021)
````

**Penjelasan:**
1. ulang_2021 = int(input("Masukan nilai batas: ")) digunakan untuk menerima nilai batas perulangan dari pengguna dan mengubahnya menjadi tipe data int.

2. jumlah_2021 = 0 digunakan sebagai nilai awal untuk menyimpan hasil penjumlahan bilangan genap.

3. for i_2021 in range(1, ulang_2021 + 1): digunakan untuk melakukan perulangan dari angka 1 sampai nilai batas yang dimasukkan.

4. if i_2021 % 2 == 0: digunakan untuk memeriksa apakah nilai i_2021 merupakan bilangan genap. Operator % digunakan untuk mendapatkan sisa hasil pembagian.

5. jumlah_2021 = jumlah_2021 + i_2021 digunakan untuk menambahkan setiap bilangan genap ke total penjumlahan.

6. Percabangan if-else di dalamnya digunakan untuk menampilkan tanda + di antara bilangan dan tanda = ketika mencapai bilangan terakhir yang diperiksa.

7. print("Jumlah =", jumlah_2021) digunakan untuk menampilkan hasil akhir penjumlahan seluruh bilangan genap.

````python
# program ini mencetak gambar jam pasir
# program ini lumayan ribet sehingga diperlukan beberapa fungsi custom

def main():
    # judul dan mengambil input user
    print("=== PROGRAM JAM PASIR KRISTAL PALINDROMIK (PEKAN 5) ===")
    n_2021 = int(input("Masukkan skala jam pasir: "))
    
    bingkai(n_2021) # lihat deklarasinya
    fase1_2021(n_2021) # lihat deklarasinya
    fase2_2021(n_2021) # lihat deklarasinya
    fase3_2021(n_2021) # lihat deklarasinya
    bingkai(n_2021) # lihat deklarasinya


# membuat bingkai jam pasir
def bingkai(n_2021):
    print("#", end="")
    for i_2021 in range (4 * n_2021 + 5):
        print("=", end="")
    print("#")


# fase 1 yang bagian jam pasir atas sebelum singularitas
def fase1_2021(n_2021):
    n_dinamis_2021 = n_2021 # variabel ini berperan sebagai acuan program untuk mencetak angka-angka

    
    # cetak seluruh baris untk bagian jam pertama alias untuk fase 1
    for i_2021 in range(n_2021):

        # garis tegak pembatas kiri dan 1 spasi padding
        print("|", end=" ")

        # spasi penyeimbang kiri sebanyak 2 * baris
        for j_2021 in range(2 * (i_2021)):
            print(" ", end="")

        # print angka dari n sampai 1    
        for j_2021 in range(n_dinamis_2021, 0, -1):
            print(f"{j_2021} ", end="")

        # poros kristal tengah karakter "<*>"
        print("<*>", end=" ")

        # print angka dari 1 sampai n
        for j_2021 in range(1, n_dinamis_2021 + 1):
            print(f"{j_2021} ", end="")

        # spasi penyeimbang kanan sebanyak 2 * baris
        for j_2021 in range(2 * (i_2021)):
            print(" ", end="")

        # garis tegak pembatas kanan
        print("|")    
        
        n_dinamis_2021 -= 1 # variabel ini berkurang 1 tiap baris
        

# fase 2 / singularitas yang mencetak "<*>" tunggal
def fase2_2021(n_2021):

    # garis tegak pembatas kiri dan 1 spasi padding
    print("|", end=" ")

    # spasi-spasi penyeimbang kiri sebanyak N
    for i_2021 in range(n_2021 * 2):
        print(" ", end="")

    # kristal tunggal
    print("<*>", end="")

    # spasi-spasi penyeimbang kanan sebanyak N
    for i_2021 in range(n_2021 * 2):
        print(" ", end="")

    # garis tegak pembatas kanan dan 1 spasi padding
    print(" ", end="|")
    print() # masuk baris baru


# fase 3 yang mencetak bagian setengah bawah jam pasir
def fase3_2021(n_2021):
    n_dinamis_2021 = 1 # variabel ini berperan sebagai acuan program untuk mencetak angka-angka
    
    # cetak seluruh baris untk bagian jam ketiga alias untuk fase 3
    for i_2021 in range(1, n_2021 + 1):

        # garis tegak pembatas kiri dan 1 spasi padding
        print("|", end=" ")

        # spasi penyeimbang kiri sebanyak 2 * (N - baris)
        for j_2021 in range(2 * (n_2021 - i_2021)):
            print(" ", end="")

        # print angka dari n sampai 1    
        for j_2021 in range(n_dinamis_2021, 0, -1):
            print(f"{j_2021} ", end="")

        # poros kristal tengah karakter "<*>"
        print("<*>", end=" ")

        # print angka dari 1 sampai n
        for j_2021 in range(1, n_dinamis_2021 + 1):
            print(f"{j_2021} ", end="")

        # spasi penyeimbang kanan sebanyak 2 * (N - baris)
        for j_2021 in range(2 * (n_2021 - i_2021)):
            print(" ", end="")

        # garis tegak pembatas kanan
        print("|")    
        
        n_dinamis_2021 += 1 # variabel ini berkurang 1 tiap baris

if __name__ == "__main__":
    main()
````

3. `fase1_2021(n_2021)` digunakan untuk membuat bagian atas jam pasir. Perulangan mencetak spasi, angka yang menurun dari `n` ke `1`, simbol `<*>`, kemudian angka yang meningkat dari `1` ke `n`. Nilai `n_dinamis_2021` dikurangi satu pada setiap baris sehingga pola semakin menyempit.

4. `fase2_2021(n_2021)` digunakan untuk membuat bagian tengah atau **singularitas** jam pasir. Bagian ini hanya mencetak satu simbol `<*>` yang berada di tengah dan diapit oleh spasi.

5. `fase3_2021(n_2021)` digunakan untuk membuat bagian bawah jam pasir. Cara kerjanya mirip dengan `fase1_2021`, tetapi `n_dinamis_2021` dimulai dari `1` dan bertambah satu pada setiap baris sehingga pola kembali melebar.

6. Perulangan `for` dan parameter `end=""` digunakan untuk mengatur posisi setiap karakter pada baris, sedangkan spasi digunakan untuk membentuk simetri pola.

7. `if __name__ == "__main__":` digunakan untuk memastikan fungsi `main()` dijalankan ketika file Python tersebut dijalankan secara langsung.
